## Ajustes na Idade

```python
df['Idade'].dtype
```

```python
# 1. Garante que os dados são numéricos (Limpa textos ocultos e transforma em float temporariamente)
df['Idade'] = pd.to_numeric(df['Idade'], errors='coerce')

# 2. Cria a coluna filtro (Idades válidas entre 13 e 90)
df['Idade_Filtro'] = df['Idade'].where((df['Idade'] >= 13) & (df['Idade'] <= 90))

# 3. Função corrigida e segura para calcular a moda
def calcular_moda_cliente(series):
    # Pega apenas o que for maior/igual a 13 e menor/igual a 90
    idades_validas = series[(series >= 13) & (series <= 90)]
    if not idades_validas.empty:
        return idades_validas.mode()[0]
    return np.nan

# 4. Aplica a moda por cliente
moda_por_cliente = df.groupby('ID_do_cliente')['Idade_Filtro'].transform(calcular_moda_cliente)
df['Idade'] = df['Idade_Filtro'].fillna(moda_por_cliente)

# 5. Segurança máxima: Se ainda sobrar algum nulo, usa a moda geral da base inteira
idades_gerais_validas = df['Idade_Filtro'][(df['Idade_Filtro'] >= 13) & (df['Idade_Filtro'] <= 90)]
if not idades_gerais_validas.empty:
    moda_geral = idades_gerais_validas.mode()[0]
    df['Idade'] = df['Idade'].fillna(moda_geral)

# 6. Preenche qualquer nulo restante com uma idade padrão (ex: 30) e converte para inteiro
df['Idade'] = df['Idade'].fillna(30).astype(int)

# 7. Remove a coluna de suporte
df = df.drop(columns=['Idade_Filtro'])
```

## Outras Substituições

```python
def substituir_por_moda(df, coluna, condicao_invalida):
    coluna_com_nan = df[coluna].where(~condicao_invalida)

    moda_por_cliente = df.groupby("ID_do_cliente")[coluna].transform(
        lambda x: x[~condicao_invalida].mode()[0]
        if not x[~condicao_invalida].mode().empty
        else np.nan
    )

    df[coluna] = df[coluna].where(~condicao_invalida, moda_por_cliente)
```

```python
# A) -100 na coluna 'Número_de_empréstimo'
substituir_por_moda(df, "Número_de_empréstimo", df["Número_de_empréstimo"] == -100)

# B) Negativos na coluna 'Dias_de_atraso_após_vencimento'
substituir_por_moda(
    df, "Dias_de_atraso_após_vencimento", df["Dias_de_atraso_após_vencimento"] < 0
)

# D) Acima de 180 na coluna 'Número_de_consultas_crédito'
substituir_por_moda(
    df, "Número_de_consultas_crédito", df["Número_de_consultas_crédito"] > 180
)

# E) Acima de 14511 na coluna 'Total_da_parcela_mensal'
substituir_por_moda(
    df, "Total_da_parcela_mensal", df["Total_da_parcela_mensal"] > 14511
)
```
